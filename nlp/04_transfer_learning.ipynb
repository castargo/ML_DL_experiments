# Transfer Learning

**Transfer Learning** — это идея "переноса" знаний из одной задачи в другую. Например, некоторая модель была обучена для задачи А, а мы хотим повторно использовать её (сделать fine‑tuning) для задачи Б.

## Типы Transfer Learning

- **Transductive** — задача одинаковая
  - Domain adaptation — разные домены
  - Cross-lingual learning — разные языки
- **Inductive** — задачи разные
  - Multi‑task learning — одновременное обучение задач
  - **Sequential Transfer Learning** — последовательное обучение задач

## Эволюция идеи

Первая попытка перенести знания - **Word Embeddings**: простейший перенос знаний - дообучение Word Embeddings под свой домен.

Перенос знаний через модель эволюционировал следующим образом: Word2Vec, GloVe ==> CoVe, ELMo ==> GPT, BERT

Такой переход стал возможен при помощи 2-х замечательных идей:
1. Переход от слов к Words‑in‑Context
    - Вместо представления отдельных слов мы можем научиться представлять слова вместе с контекстом, в котором они используются.
    - Было **Word Embeddings**: Word2Vec, GloVe
    - Стало **Word Contexts**: CoVe, ELMo, GPT, BERT
2. Отказ от моделей, ориентированных на конкретную задачу
    - Было **task‑specific models**: Word2Vec, GloVe, CoVe, ELMo
    - Стало **nonspecific models**: GPT, BERT

### Переход от слов к Words‑in‑Context

В концепции Word Embeddings каждое слово является независимым. Подходы Word Contexts кодируют слова внутри предложения. Важно: вне зависимости от того, как мы представляем слово (Word Embeddings или Word Contexts), эти представления все еще подаются в **task‑specific model**.

Подходы Word Contexts:
* CoVe (COntext VEctors): Contextualized Word Vectors Learned in Translation
    * [Learned in Translation: Contextualized Word Vectors (2017)](https://arxiv.org/pdf/1708.00107.pdf)
* ELMo: Embeddings from Language Models
    * [Deep contextualized word representations (2018)](https://arxiv.org/pdf/1802.05365.pdf)
    * ELMo - это языковая модель
* ULM-FiT

### Отказ от моделей, ориентированных на конкретную задачу

CoVe/ELMo - это замена word embeddings, но GPT/BERT могут заменить собой модель целиком. 

Представления ELMo/CoVe в основном использовались для замены embedding layer и сохраняли почти нетронутой архитектуру модели для конкретных задач (task-specific). Для каждого типа задачи (QA, part-of-speech tagging и тд) нужна была отдельная архитектура.

GPT/BERT - универсальная модель для многих задач. Модель сначала предварительно обучается с использованием огромного количества неразмеченных данных (обычного текста). Затем эта модель дорабатывается для каждой из последующих задач. Что важно, теперь во время fine-tuning вам нужно использовать только входные преобразования с учетом задач (т. е. подавать данные определенным образом) вместо изменения архитектуры модели.
