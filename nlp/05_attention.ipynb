## Attention

**Attention** — это механизм, который позволяет модели динамически фокусироваться на наиболее релевантных частях входной последовательности при генерации каждого элемента выходной. В encoder-decoder архитектуре attention решает ключевую проблему: традиционный энкодер сжимает всю входную последовательность в один фиксированный контекстный вектор, который затем используется декодером. Это плохо работает для длинных последовательностей, так как вся информация «сдавливается» в одном векторе, и декодер вынужден полагаться на одно и то же представление на каждом шаге генерации.

### Зачем и когда придумали attention

До появления attention стандартным подходом к seq2seq-задачам (например, машинный перевод) было использование двух RNN/LSTM: encoder и decoder части. В этой модели энкодер читал исходное предложение, а его финальное состояние использовалось как начальное состояние декодера. Предполагалось, что это финальное состояние способно «закодировать» всю информацию об исходном предложении, и декодер сможет сгенерировать перевод на его основе.

Такая модель с несколькими слоями была использована, например, в одной из первых работ по нейросетевым seq2seq-моделям — статье **"Sequence to Sequence Learning with Neural Networks"** (Sutskever et al., 2014). Авторы визуализировали последнее состояние кодера и обнаружили, что представления предложений со схожим смыслом, но разной структурой, оказываются близкими в этом пространстве. Однако у этого подхода был фундаментальный недостаток: вся информация о входной последовательности, какой бы длинной она ни была, должна была быть сжата в один вектор фиксированной размерности. Это создавало bottleneck и ограничивало качество перевода, особенно для длинных предложений.

Именно для решения этой проблемы был предложен механизм **attention**. Ключевая идея заключается в том, чтобы на каждом шаге декодер мог «заглядывать» во все скрытые состояния энкодера, а не полагаться только на одно финальное. Это позволяет модели динамически выбирать, какая часть исходного предложения наиболее важна для генерации текущего слова перевода.

Зачем нужен attention:
- Улучшение качества seq2seq-задач для длинных последовательностей - модель может напрямую «обращаться» к нужным частям входа, что особенно важно для длинных предложений.
- Решение bottelbeck проблемы - модель больше не ограничена фиксированным вектором и может использовать всю информацию из входной последовательности.

### Как работает attention

Вместо одного фиксированного контекстного вектора attention позволяет декодеру на каждом шаге **вычислять** контекстный вектор заново как **взвешенную сумму** всех скрытых состояний кодера. Веса attention определяют, насколько каждый входной токен важен для генерации текущего выходного токена.

Процесс:

1. **Вычисление оценок (scores)**: Для текущего скрытого состояния декодера $h_t$ и каждого скрытого состояния кодера $\bar{h}_s$ вычисляется оценка $e_{t,s} = a(h_t, \bar{h}_s)$. Функция $a$ может быть, например, скалярным произведением (dot product) или обучаемой нейронной сетью (как в Bahdanau model).

2. **Нормализация (softmax)**: Оценки $e_{t,s}$ преобразуются в веса $\alpha_{t,s}$ с помощью функции softmax, чтобы они в сумме давали 1:

$$\alpha_{t,s} = \frac{\exp(e_{t,s})}{\sum_{k=1}^{T_x} \exp(e_{t,k})}$$

3. **Вычисление контекстного вектора**: Контекстный вектор $c_t$ для текущего шага $t$ вычисляется как взвешенная сумма всех скрытых состояний кодера:

$$c_t = \sum_{s=1}^{T_x} \alpha_{t,s} \bar{h}_s$$

4. **Использование контекстного вектора**: Контекстный вектор $c_t$ объединяется со скрытым состоянием декодера $h_t$ (например, конкатенируется) и используется для предсказания следующего токена.

### Модель Bahdanau

Первая модель, в которой был применён механизм attention, получила название **Bahdanau model** (или **additive attention**) — по имени первого автора статьи **"Neural Machine Translation by Jointly Learning to Align and Translate"** (Bahdanau et al., 2014). В этой работе было предложено устранить "бутылочное горлышко" encoder-decoder архитектуры для машинного перевода при помощи весов, присваемых промежуточным скрытым значениям. Такие веса позволяют задать количество "внимания", которое модель должна уделять входным данным на каждом этапе декодирования.

Ключевая идея Bahdanau model заключается в следующем:
- Энкодер — bidirectional RNN, который читает исходное предложение в обоих направлениях и сохраняет все скрытые состояния.
- Декодер — RNN, которая на каждом шаге генерации вычисляет контекстный вектор как взвешенную сумму всех скрытых состояний кодера.
- Веса внимания вычисляются с помощью feedforward network, которая принимает на вход текущее скрытое состояние декодера и каждое скрытое состояние энкодера.

Благодаря этому механизму модель может «выравнивать» слова исходного предложения и перевода, что значительно улучшает качество перевода, особенно для длинных предложений. Визуализация весов внимания в Bahdanau model показала, что модель действительно учится правильно сопоставлять слова в разных языках, что делает её интерпретируемой.

### Модель Luong

**Luong model** (или **multiplicative attention**) была предложена в статье **"Effective Approaches to Attention-based Neural Machine Translation"** (Luong et al., 2015) как более простая и вычислительно эффективная альтернатива Bahdanau model.

Ключевые отличия Luong model:

1. **Использование текущего состояния декодера**: В Bahdanau attention вычисляется на основе скрытого состояния декодера с предыдущего шага ($h_{t-1}$). В Luong attention используется скрытое состояние текущего шага ($h_t$). Сначала декодер генерирует предварительное состояние, а затем, используя его, вычисляется контекстный вектор.

2. **Более простые функции выравнивания (score functions)**: Вместо обучаемой нейросети (как в Bahdanau), Luong предлагает три простых способа вычисления оценок:
   - **Dot (скалярное произведение)**: $\text{score}(h_t, \bar{h}_s) = h_t^T \bar{h}_s$
   - **General (обобщённое)**: $\text{score}(h_t, \bar{h}_s) = h_t^T W_a \bar{h}_s$, где $W_a$ — обучаемая матрица весов
   - **Concat (конкатенация)**: $\text{score}(h_t, \bar{h}_s) = v_a^T \tanh(W_a [h_t; \bar{h}_s])$ — похоже на подход Богданау, но в упрощённом виде

3. **Два подхода к вниманию: Global и Local**:
   - **Global Attention**: Модель обращается ко всем скрытым состояниям кодера на каждом шаге (аналог Bahdanau).
   - **Local Attention**: Модель для каждого шага предсказывает позицию во входной последовательности и фокусируется только на небольшом окне контекста вокруг неё. Это значительно ускоряет вычисления и улучшает качество на длинных последовательностях.

Благодаря использованию скалярного произведения как основной функции выравнивания, Luong Attention требует меньше вычислений и параметров, чем аддитивное внимание Богданау. Эта архитектура стала важным шагом на пути к Transformer'у, который, по сути, использует глобальное Luong Attention (в виде масштабированного скалярного произведения) как один из своих ключевых строительных блоков.

> Luong предложил три способа вычисления оценок: dot, general и concat. Все они относятся к multiplicative attention (потому что используют умножение матриц).

### Семейство механизмов внимания

[В статье Lilian Weng](https://lilianweng.github.io/posts/2018-06-24-attention/) приводится сводная таблица популярных механизмов внимания и соответствующих им функций оценки (alignment score functions).

| Название механизма | Функция оценки (Alignment Score) | Источник |
| :--- | :--- | :--- |
| **Content-base attention** | $\text{score}(s_t, h_i) = \text{cosine}[s_t, h_i]$ | Graves 2014 |
| **Additive attention** | $\text{score}(s_t, h_i) = v_a^\top \tanh(W_a [s_{t-1}; h_i])$ | Bahdanau 2015 |
| **Location-Base** | $\alpha_{t,i} = \text{softmax}(W_a s_t)$ | Luong 2015 |
| **General** | $\text{score}(s_t, h_i) = s_t^\top W_a h_i$ | Luong 2015 |

| **Scaled Dot-Product** | $\text{score}(s_t, h_i) = \frac{s_t^\top h_i}{\sqrt{n}}$ | Vaswani 2017 |

## Self-Attention

**Self-attention** — это механизм, который позволяет каждому элементу последовательности «взаимодействовать» со всеми остальными элементами этой же последовательности, вычисляя для себя новое представление, учитывающее контекст всей последовательности.

В отличие от attention в seq2seq, где внимание вычисляется между двумя разными последовательностями (энкодер и декодер), self-attention работает внутри одной последовательности энкодера: **запросы (queries)**, **ключи (keys)** и **значения (values)** берутся из одной и той же последовательности.

### Attention is All You Need

Ключевой момент в истории self-attention — статья [**«Attention Is All You Need»**](https://arxiv.org/abs/1706.03762) (Vaswani et al., 2017). В ней авторы предложили архитектуру **Transformer**, которая **полностью отказалась от рекуррентных (RNN) и свёрточных (CNN) слоёв**. По аналогии с Encoder-Decoder архитектурой состовляющие Transformer-а можно разложить на следующие составляющие:
* Encoder: Multi-Head Attention - каждое слово в входном предложении «смотрит» на все остальные слова этого же предложения.
* Decoder: Masked Multi-Head Attention - при генерации выходного слова декодер может «смотреть» только на ранее сгенерированные слова (используется маска, чтобы не заглядывать в будущее).
* Связь между encoder и decoder частями: Multi-Head Attention - запросы приходят из декодера, а ключи и значения — из энкодера.

### Как работает self-attention

$X$ — матрица входных эмбеддингов:

$$
X = 
\begin{bmatrix}
h_1^T \\
h_2^T \\
\vdots \\
h_n^T
\end{bmatrix} \in \mathbb{R}^{n \times d_{\text{model}}}
$$

$ W_Q, W_K, W_V \in \mathbb{R}^{d_{\text{model}} \times d_k} $  — обучаемые матрицы параметров.

Они переводят вектора $h_i$ из $ \mathbb{R}^{d_{\text{model}}} $ в вектора $ q_i, k_i, v_i $ из $ \mathbb{R}^{d_k} $, а матрицу $X$ размерности $ n \times d_{\text{model}} $ в матрицы $ Q, K, V $ размерности $ n \times d_k $.

#### **Линейные преобразования последовательности $X$**

- Для каждого элемента последовательности вычисляются три вектора:
    - **Запрос (Query)**: $Q = X W_Q$
    - **Ключ (Key)**: $K = X W_K$
    - **Значение (Value)**: $V = X W_V$
    - Матрицы Q и K должны иметь одну размерность, матрица V может иметь любую размерность, которая станет размерностью выходного контекстного вектора.
- Роли векторов:
    - **Запрос (Q)** — "какую информацию я ищу?" (запрос от текущего элемента)
    - **Ключ (K)** — "какая информация у меня есть?" (описание всех элементов, на которые можно обратить внимание)
    - **Значение (V)** — "что я передам дальше?" (содержание, которое будет извлечено)

#### **Вычисление оценок attention**

В оригинальном Transformer используется scaled dot-product attention:
- Для каждой пары элементов вычисляется оценка, показывающая, насколько запрос одного элемента «соответствует» ключу другого.
- По сути нужно проверить, насколько каждый токен похож (важен) на каждый токен, то есть взять их скалярное произведение (в матричном виде для всех токенов всех последовательностей: $Q K^T$).
- Сами по себе скалярные произведения сложны к интерпретации, потому что могут принимать очень маленькие или очень большие значения. Поэтому будем использовать softmax для получения матрицы нормированных весов.
- Деление на $\sqrt{d_k}$ необходимо для стабилизации градиентов. При больших размерностях скалярные произведения становятся очень большими, что приводит к экстремально малым градиентам после softmax. $d_k$ — размерность ключей.
- Выходное представление для каждого элемента вычисляется как взвешенная сумма значений (V) всех элементов.

Итоговая формула выглядит так:

$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{Q K^T}{\sqrt{d_k}}\right) V$$

На выходе мы получаем тезор размера: (seq_len, batch, d_model).

### Почему self-attention лучше RNN и CNN

В оригинальной статье авторы приводят сравнительный анализ трёх архитектур:

| Характеристика | Self-Attention | RNN | CNN |
|----------------|----------------|--------|--------|
| **Сложность на слой** | $O(n^2 \cdot d)$ | $O(n \cdot d^2)$ | $O(k \cdot n \cdot d^2)$ |
| **Минимальное число последовательных операций** | $O(1)$ | $O(n)$ | $O(1)$ |
| **Максимальный путь для дальних зависимостей** | $O(1)$ | $O(n)$ | $O(\log_k n)$ |

Ключевые преимущества self-attention:

- **$O(1)$ последовательных операций**: Все позиции обрабатываются одновременно.
- **Прямой доступ к любой позиции**: Расстояние между любыми двумя словами всегда равно одному шагу.
- **Идеально для GPU/TPU**: Высокий параллелизм позволяет эффективно использовать современное оборудование.

### Позиционное кодирование (Positional Encoding)

Self-attention не понимает порядок слов, а также обрабатывает их не последовательно, а параллельно. Чтобы модель понимала порядок слов, в Transformer добавляется **позиционное кодирование** — вектор, который добавляется (не конкатенируется!) к эмбеддингу каждого слова.

Итоговый вектор для токена $i$ - это $h_i = x_i + p_i$. Такое преобразование не меняет форму тензора, но позволяет закодировать информацию не только о содержании слова, но и о его позиции.

В оригинальной статье используется синусоидальное кодирование:

$$PE_{(pos, 2i)} = \sin(pos / 10000^{2i/d_{\text{model}}})$$
$$PE_{(pos, 2i+1)} = \cos(pos / 10000^{2i/d_{\text{model}}})$$

Это кодирование обладает важным свойством: для любого фиксированного смещения $k$ позицию $pos+k$ можно выразить как линейную функцию от позиции $pos$.

### Masked Self-attention

В декодере Transformer self-attention модифицируется с помощью маскирования, чтобы предотвратить «подглядывание» в будущие токены при генерации. Это необходимо, чтобы модель не могла использовать информацию о словах, которые ещё не были сгенерированы, что гарантирует авторегрессивный характер генерации.

**Как работает маскирование:**

1. Перед вычислением softmax к оценкам внимания (scores) добавляется маска, которая обнуляет (заменяет на $-\infty$) все позиции, соответствующие будущим токенам.
2. После применения softmax веса для будущих позиций становятся равными нулю, и они не влияют на взвешенную сумму значений (V).

Это позволяет модели обрабатывать всю последовательность параллельно, но при этом каждое слово «видит» только себя и предыдущие слова.

**Формула masked attention:**

$$\text{MaskedAttention}(Q, K, V) = \text{softmax}\left(\frac{Q K^T}{\sqrt{d_k}} + M\right) V$$

где $M$ — матрица маски, содержащая 0 для разрешённых позиций и $-\infty$ для запрещённых.

Маска обычно имеет вид верхней треугольной матрицы (для последовательности длины $n$), где все элементы выше главной диагонали равны $-\infty$.

Применение:
- Во время обучения маска применяется на каждом шаге, чтобы модель училась предсказывать следующий токен, не видя правильный ответ.
- Во время инференса (генерации) маска всё ещё используется, чтобы при пошаговой генерации модель опиралась только на уже сгенерированные токены.

### Интуитивное понимание self-attention

Чтобы понять, зачем нужны три проекции — Query, Key, Value — и как они работают, можно воспользоваться аналогией с **нечётким словарём**.

Представьте обычный словарь (hash map): `{'Ann': 167, 'Bob': 182, 'Zoe': 192}`. Здесь ключи — имена людей, значения — их рост. Если мы делаем запрос (Query) `"Ann"`, то получаем значение `167`. Если запрос `"Carl"` — ничего не получаем, потому что такого ключа нет. А что, если запрос `"Anna"`? Это почти `"Ann"`, но точного совпадения нет. В стандартном словаре мы бы ничего не получили, но логически мы можем предположить, что рост `"Anna"` близок к росту `"Ann"`.

Идея **внимания** — взять **взвешенное среднее** значений (Value) по всем ключам, где веса определяются **похожестью** запроса (Query) на каждый ключ (Key). Например, для запроса `"Anna"` мы можем дать вес 0.96 ключу `"Ann"`, и по 0.02 ключам `"Bob"` и `"Zoe"`. Тогда ответом будет: `0.96 * 167 + 0.02 * 182 + 0.02 * 192 ≈ 168`. Это и есть нечёткий поиск в словаре.

Теперь усложним: пусть ключи и запросы — это **векторы признаков**, например, данные о компаниях: выручка в прошлом году, число сотрудников и т.д. Значения (Value) — это выручка в этом году (число). Тогда словарь выглядит так:
`{ [17, 20000, ...]: 400, [11, 10000, ...]: 100, ... }`.

Если поступает новая компания с вектором признаков `[13, 15000, ...]`, мы можем оценить её выручку, сравнив её вектор с векторами ключей (например, через косинусное сходство или скалярное произведение). Полученные веса используем для взвешенного суммирования значений. Это в точности то, что делает метод **k ближайших соседей (kNN) в задаче регрессии** — только вместо выбора ближайших соседей мы используем все, но с разными весами.

Наконец, в задачах NLP значения (Value) — это не числа, а **векторы** (эмбеддинги токенов). В self-attention каждый токен (слово) может обновить своё представление, «посмотрев» на все остальные токены. Для этого мы превращаем каждый токен в три вектора: Query (что ищу), Key (что могу дать), Value (что передам). Далее для каждого токена вычисляем его сходство с ключами всех токенов, получаем веса и взвешиваем соответствующие значения. В результате каждый токен получает новое представление, обогащённое контекстом всей последовательности.

Математически это записывается как **скалярное произведение**:

$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{Q K^T}{\sqrt{d_k}}\right) V$$

Здесь $Q$ — матрица запросов, $K$ — ключей, $V$ — значений, а $d_k$ — размерность ключей (деление на $\sqrt{d_k}$ нужно для стабилизации градиентов).

Выход внимания для i-го токена:

$$Z_i = \sum_{j=1}^n A_{ij} v_j = \sum_{j=1}^n \frac{\exp(q_i k_j)}{\sum_{t=1}^n \exp(q_i k_t)} v_j$$

```python
class ScaleDotProductAttention(nn.Module):
    """
    compute scale dot product attention

    Query : given sentence that we focused on (decoder)
    Key : every sentence to check relationship with Qeury(encoder)
    Value : every sentence same with Key (encoder)
    """

    def __init__(self):
        super(ScaleDotProductAttention, self).__init__()
        self.softmax = nn.Softmax(dim=-1)

    # tokens [B, L, D] -> [B, L, d_tensor*H] -> [B, L, d_tensor, H] -> [B, L, H, d_tensor]
    def forward(self, q, k, v, mask=None):
        # input is 4 dimension tensor
        batch_size, head, length, d_tensor = k.size()
        
        # q (B, H, L, D) 
        # k (B, H, L, D) 
        # v (B, H, L, D) 
        
        # L -> LxL [prefix, l1, l2, l3, l4] -> [l1, l2, l3, l4, <EOS>] -> 5x5 
        # -> [1, 0, 0, 0, 0] / [x, y, 0, 0, 0] / [x, y, z, 0, 0] / [x, y, z, x1, 0] / [x, y, z, x1, y1]
        
        # self_attention = softmat((q@k.T)/sqrt(d_tensor))@v LxL Lxd_v -> Lxd_v

        k_t = k.transpose(-1, -2)  # transpose  k_t (B, L, H, D)
        score = (q @ k_t) / torch.sqrt(d_tensor) # (B, H, L, D) x (B, H, D, L) (B, H, L, L)

        if mask is not None:
            score = score.masked_fill(mask == 0, float("-inf"))
            
        score = self.softmax(score)
        v = score @ v
        
        return v
```

## Multi-Head Attention

**Multi-head attention** — это механизм, который запускает несколько "голов" attention параллельно, чтобы модель могла смотреть на последовательность под разными углами. Вместо одного «внимания» модель использует $H$ параллельных «голов» attention, каждая из которых вычисляет своё собственное представление.

Внутренний размер модели мы задавали как $d_{model}$. Для Multi-head attention необходимо выбрать $d_{model}$ так, чтобы его значение было кратно числу голов $H$. Тогда каждая голова будет работать в подпространстве размерности $\frac{d_{model}}{H}$

### Как работает multi-head attention

1. **Линейное проецирование**: Входная последовательность $X$ проецируется в $H$ разных наборов $Q$, $K$ и $V$ с помощью разных весовых матриц:

   $$Q_i = X W_Q^i, \quad K_i = X W_K^i, \quad V_i = X W_V^i$$

   для каждой головы $i = 1, \dots, h$.

2. **Параллельное вычисление self-attention**: Для каждой головы независимо вычисляется self-attention:

   $$\text{head}_i = \text{Attention}(Q_i, K_i, V_i)$$
   
   При этом мы разбиваем каждый тензор на $H$ голов, например, новая размерность $Q$: (batch, $H$, seq_len, $d_k$).

3. **Конкатенация и линейное преобразование**: Все выходы голов конкатенируются и проецируются с помощью ещё одной весовой матрицы $W_O$:

   $$\text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, \dots, \text{head}_h) W_O$$
  
   При этом для каждой головы:
   
   $$\text{scores}_i = \frac{Q_i K_i^T}{\sqrt{\frac{d_{\text{model}}}{H}}}, \quad \text{weights}_i = \text{softmax}(\text{scores}_i)$$
   
   $$H_i = \text{weights}_i V_i$$

### Зачем нужен multi-head attention

- **Разные «взгляды» на данные**: Каждая голова может учиться фокусироваться на разных типах отношений. Например, одна голова может отслеживать синтаксические зависимости, другая — семантические, третья — позиционные. Исследования показывают, что большинство голов можно удалить без значительной потери качества, но в процессе обучения они необходимы для изучения этих полезных функций.
- **Улучшенное представление**: Конкатенация выходов разных голов позволяет модели создавать более богатое и информативное представление последовательности.
- **Гибкость**: Количество голов можно настраивать в зависимости от задачи и размера модели. В оригинальном Transformer используется 8 голов, но в больших моделях их может быть значительно больше.

### Связь с анализом голов attention

Исследования показывают, что даже в обученной модели многие головы attention можно «обрезать» (prune) без потери качества, что подтверждает наличие специализированных, наиболее важных голов. Это также говорит о том, что multi-head attention — это не просто избыточный механизм, а скорее способ дать модели достаточно «пространства» для изучения разнообразных паттернов, которые затем могут быть сжаты.

## Cross‑Attention (Encoder‑Decoder Attention)

Cross‑attention — это механизм, где **запросы (Q)** приходят из одной последовательности (обычно из декодера), а **ключи (K) и значения (V)** — из другой (из энкодера). Это позволяет декодеру на каждом шаге генерации «обращаться» к входной последовательности и выбирать наиболее релевантные части.

**Где используется:**
- В классическом Transformer между энкодером и декодером.
- В мультимодальных моделях (текст → изображение, изображение → текст).
- В задачах типа «вопрос-ответ», где контекст (документ) подаётся как K/V, а вопрос — как Q.

## Трюки для вычисления attention

Стандартный механизм self-attention, лежащий в основе Transformer, обладает вычислительной сложностью $O(n^2)$ как по времени, так и по памяти (где $n$ — длина последовательности). Это ограничение становится критическим при работе с длинными текстами и при обучении больших моделей.

### Sparse Attention

**Идея:** вместо того чтобы каждый токен «общался» с каждым, разрешить ему взаимодействовать только с подмножеством токенов. Это снижает сложность с квадратичной до линейной ($O(n)$) или близкой к ней.

Ключевые модели, реализующие этот подход:

*   **Longformer (Beltagy et al., 2020)**: Использует комбинацию двух паттернов внимания:
    1.  **Sliding Window (локальное окно)**: каждый токен attends только к соседям в пределах окна фиксированного размера (например, 512 токенов). Это основа, которая покрывает локальный контекст.
    2.  **Global Attention (глобальные токены)**: отдельные токены (например, `[CLS]`) имеют доступ ко всей последовательности и наоборот. Это позволяет модели захватывать глобальный контекст, необходимый для понимания документа в целом.

*   **BigBird (Zaheer et al., 2020)**: Развивает идею Longformer, добавляя третий паттерн:
    1.  **Window Attention** (локальное окно).
    2.  **Global Attention** (глобальные токены).
    3.  **Random Attention** (случайное внимание): каждый токен attends к нескольким случайным токенам во всей последовательности.

Такая комбинация позволяет BigBird не только эффективно обрабатывать длинные последовательности (до 8× длиннее, чем у стандартного Transformer), но и **теоретически сохранять свойства full-внимания** (быть «универсальным аппроксиматором»), что не всегда гарантировано для других разреженных методов.

> Sparse Attention — это **изменение архитектуры внимания**, которое позволяет обрабатывать значительно более длинные последовательности, жертвуя полнотой связей между токенами ради вычислительной эффективности.

### Flash Attention

Flash Attention — это алгоритм, который ускоряет вычисление attention и снижает потребление памяти. Он был предложен в работе **"FlashAttention: Fast and Memory-Efficient Exact Attention with IO-Awareness"** (Dao et al., 2022).

**Основная идея:**
Стандартное внимание требует чтения и записи всей матрицы весов $n \times n$ в медленную память GPU (HBM). Flash Attention использует **разбиение на блоки (tiling)**: матрицы Q, K, V разбиваются на небольшие блоки, которые помещаются в быструю кэш-память (SRAM). Вычисления производятся поблочно, а финальный результат пересчитывается **on-the-fly** без сохранения полной матрицы весов.

**Преимущества:**
* **Скорость:** Ускорение в **2–4 раза** по сравнению со стандартной реализацией.
* **Память:** Сложность по памяти снижается с $O(n^2)$ до **линейной** $O(n)$.
* **Точность:** В отличие от Sparse Attention, это **точное** внимание — результат идентичен стандартному.

**Влияние на индустрию:**
Flash Attention стала одной из ключевых технологий, позволивших перейти к моделям с контекстным окном в **128K, 1M и более токенов**. Без неё обучение и инференс таких моделей были бы экономически нецелесообразны. Она широко используется в современных LLM, таких как GPT-4, LLaMA и Mistral.

> Flash Attention — это **оптимизация вычислений на уровне GPU**, которая делает стандартное внимание быстрее и экономичнее по памяти, не меняя его математической сути.
